## BFS WITH QUEUE

```python
import collections

# BFS with queue (using deque for efficient FIFO)

# Using a tree to perform BFS
tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs_with_queue(tree, start_node, goal_node):
    visited = set()
    bfs_queue = collections.deque([start_node])  # Using deque as queue (FIFO)
    
    while bfs_queue:
        next_node = bfs_queue.popleft()  # Remove first element (FIFO)
        
        if next_node not in visited:  # Process node only if not visited
            print(next_node, end=" ")
            visited.add(next_node)
            
            if next_node == goal_node:  # If goal node is reached
                print("Goal Reached!")
                return visited
            
            for child in tree[next_node]:  # Add children to the end of the queue
                if child not in visited:
                    bfs_queue.append(child)
    
    return visited

visited_nodes = bfs_with_queue(tree, 'A', 'G')  # Start BFS from 'A' searching for 'G'
```

## BFS WITHOUT QUEUE

```python
# BFS without queue (using list as FIFO)

# Using a tree to perform BFS
tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs_without_queue(tree, start_node, goal_node):
    visited = set()
    bfs_list = [start_node]  # using list as a queue (FIFO)
    
    while bfs_list:
        next_node = bfs_list.pop(0)  # Remove first element (FIFO)
        
        if next_node not in visited:  # Process node only if not visited
            print(next_node, end=" ")
            visited.add(next_node)
            
            if next_node == goal_node:  # If goal node is reached
                print("Goal Reached!")
                return visited
            
            for child in tree[next_node]:  # Add children to the end of the list
                if child not in visited:
                    bfs_list.append(child)
    
    return visited

visited_nodes = bfs_without_queue(tree, 'A', 'G')  # Start BFS from 'A' searching for 'G'
```
