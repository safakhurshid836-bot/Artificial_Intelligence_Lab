Let's create a sample DataFrame with missing values to illustrate `ffill` and `bfill`.

```
import pandas as pd
import numpy as np

sample_data = {'A': [1, 2, np.nan, 4, 5],
               'B': [np.nan, 7, 8, np.nan, 10],
               'C': [11, 12, 13, 14, np.nan]}
sample_df = pd.DataFrame(sample_data)

print("Original DataFrame with missing values:")
display(sample_df)
```

### Using Forward Fill (`ffill`)

```
df_ffill = sample_df.ffill()
print("DataFrame after forward fill (ffill):")
display(df_ffill)
```

### Using Backward Fill (`bfill`)

```
df_bfill = sample_df.bfill()
print("DataFrame after backward fill (bfill):")
display(df_bfill)
```

Other common methods for handling missing values include:

*   **Filling with a specific value**: `df.fillna(value)` (e.g., `df.fillna(0)` to replace NaNs with 0).
*   **Filling with the mean, median, or mode**: `df['column'].fillna(df['column'].mean())`.
*   **Dropping rows/columns with missing values**: `df.dropna()`.

# Task
Preprocess the `df` DataFrame by performing the following steps: check for and remove duplicate rows, identify categorical and numerical features, apply one-hot encoding to categorical features, standardize numerical features using `StandardScaler`, and then display the first few rows and information about the resulting preprocessed DataFrame. Finally, summarize all the preprocessing operations that were performed.

## Check for Duplicates

### Subtask:
Identify and remove any duplicate rows in the DataFrame to ensure data uniqueness.


**Reasoning**:
First, I'll identify and display the number of duplicate rows in the DataFrame to understand the extent of the problem. Then, I'll remove these duplicates from the DataFrame to ensure data uniqueness, and finally, I'll display the new shape of the DataFrame to confirm the duplicates have been successfully removed.



```
print(f"Number of duplicate rows before removal: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print(f"DataFrame shape after removing duplicates: {df.shape}")
```

## Identify Categorical and Numerical Features

**Reasoning**:
The next step in preprocessing is to identify which columns are categorical and which are numerical. This is a crucial step for applying appropriate preprocessing techniques later, such as one-hot encoding for categorical features and standardization for numerical features. I will use a code block to automatically classify columns based on their data types.



```
categorical_features = df.select_dtypes(include='object').columns
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns

print(f"Categorical Features ({len(categorical_features)}):\n{list(categorical_features)}")
print(f"Numerical Features ({len(numerical_features)}):\n{list(numerical_features)}")
```

## Apply One-Hot Encoding

**Reasoning**:
To prepare the categorical features for machine learning models, I will apply one-hot encoding using `pd.get_dummies`. This converts each categorical variable into a new set of binary variables, which is suitable for many algorithms.



```
df_encoded = pd.get_dummies(df, columns=categorical_features, drop_first=True)
print("DataFrame after one-hot encoding:")
display(df_encoded.head())
```

## Standardize Numerical Features

**Reasoning**:
Now that one-hot encoding has been applied, I will standardize the numerical features to ensure that they contribute equally to the model training, preventing features with larger values from dominating. I will use `StandardScaler` from `sklearn.preprocessing` for this purpose and then display the first few rows of the modified DataFrame.



```
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_encoded[numerical_features] = scaler.fit_transform(df_encoded[numerical_features])

print("DataFrame after standardizing numerical features:")
display(df_encoded.head())
```

**Reasoning**:
After standardizing numerical features, I will now display the information about the final preprocessed DataFrame to show its structure and data types after all the transformations.



```
print("Information about the preprocessed DataFrame:")
df_encoded.info()
```
