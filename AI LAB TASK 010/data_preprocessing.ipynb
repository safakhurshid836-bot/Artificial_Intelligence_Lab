```python
import pandas as pd
```

```python
exoplanet_data = pd.read_csv('cumulative.csv', sep = ',')
exoplanet_data
```

```python
clean_col_used = [
    "koi_period",
    "koi_duration",
    "koi_depth",
    "koi_prad",
    "koi_teq",
    "koi_insol",
    "koi_model_snr",
    "koi_steff",
    "koi_slogg",
    "koi_srad",
    "koi_kepmag",
    "koi_disposition"
]
```

```python
exoplanet_data = exoplanet_data[clean_col_used]
exoplanet_data
```

```python
exoplanet_data = exoplanet_data.dropna()
exoplanet_data
```

```python
exoplanet_data = exoplanet_data[exoplanet_data["koi_disposition"].isin(["CONFIRMED", "FALSE POSITIVE"])]
```

```python
exoplanet_data["koi_disposition"] = exoplanet_data["koi_disposition"].map({"CONFIRMED": 1,"FALSE POSITIVE": 0})
exoplanet_data
```

```python
exoplanet_data.info()
```

```python
exoplanet_data.to_csv('cleaned_exoplanet_data.csv', index=False, header=True)
```
